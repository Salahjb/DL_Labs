```python
url3 = "https://www.aljazeera.net/encyclopedia/2016/6/6/%D8%A7%D9%84%D8%B0%D9%83%D8%A7%D8%A1-%D8%A7%D9%84%D8%A7%D8%B5%D8%B7%D9%86%D8%A7%D8%B9%D9%8A-%D8%AD%D9%8A%D9%86%D9%85%D8%A7-%D8%AA%D9%81%D9%83%D8%B1-%D8%A7%D9%84%D8%A2%D9%84%D8%A9"

response3 = requests.get(url3)

soup3 = BeautifulSoup(response3.content , 'html.parser')
arabic_text.append(soup3.find('p').text.strip())
```

```python
url = "https://mawdoo3.com/%D8%AA%D8%B9%D8%B1%D9%8A%D9%81_%D8%A7%D9%84%D8%B0%D9%83%D8%A7%D8%A1_%D8%A7%D9%84%D8%A7%D8%B5%D8%B7%D9%86%D8%A7%D8%B9%D9%8A"

response = requests.get(url)
soup = BeautifulSoup(response.content, 'html.parser')
texts = soup.find_all('p')

arabic_texts = [p.text for p in texts if any('\u0600' <= c <= '\u06FF' for c in p.text)]
```

```python
for i, text in enumerate(arabic_texts):
    print(f"Text {i}:\n{text}\n")
```

```python
arabic_Text = arabic_text + arabic_texts
```

```python
len(arabic_Text)
```

```python
scores = []
for text in arabic_Text:
    print("\nText:", text)
    score = float(input("Assign a relevance score (0–10): "))
    scores.append(score)
```

```python
scores
```

```python
import pandas as pd

df = pd.DataFrame({
    'Text': arabic_Text,
    'Score': scores
})
```

```python
import pandas as pd
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem.isri import ISRIStemmer
from pyarabic.araby import strip_tashkeel, strip_tatweel, normalize_hamza
```

```python
# Download required NLTK resources
nltk.download('punkt')
nltk.download('stopwords')
# This is needed specifically for Arabic tokenization
nltk.download('punkt_tab')
```

```python

# Load Arabic stopwords
arabic_stopwords = set(stopwords.words("arabic"))
stemmer = ISRIStemmer()
```

```python
def preprocess_arabic(text):
    # Add error handling to handle NaN values or non-string data
    if pd.isna(text) or not isinstance(text, str):
        return ""
        
    # Remove diacritics and elongation
    text = strip_tashkeel(text)
    text = strip_tatweel(text)
    
    # Normalize Hamza and alef variations
    text = normalize_hamza(text)
    text = re.sub(r'[^\u0600-\u06FF\s]', '', text)  # Keep only Arabic characters
    
    # Tokenize - use a more reliable method for Arabic
    try:
        words = nltk.word_tokenize(text, language='arabic')
    except LookupError:
        # Fallback to simple whitespace tokenization if NLTK tokenization fails
        words = text.split()
    
    # Remove stopwords and short words
    filtered = [stemmer.stem(w) for w in words if w not in arabic_stopwords and len(w) > 2]
    
    return ' '.join(filtered)
```

```python
# processing my dataframe
try:
    df['Cleaned_Text'] = df['Text'].apply(preprocess_arabic)
    print(df[['Text', 'Cleaned_Text']].head())
except Exception as e:
    print(f"Error processing dataframe: {e}")
```
