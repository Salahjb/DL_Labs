```python
import pandas as pd
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem.isri import ISRIStemmer
from pyarabic.araby import strip_tashkeel, strip_tatweel, normalize_hamza
```

```python
# Download required NLTK resources
nltk.download('punkt')
nltk.download('stopwords')
# This is needed for Arabic tokenization nltk.download('punkt_tab')
```

```python
# Load Arabic stopwords
arabic_stopwords = set(stopwords.words("arabic"))
stemmer = ISRIStemmer()
```

```python
def preprocess_arabic(text):
    # Add error handling to handle NaN values or non-string data
    if pd.isna(text) or not isinstance(text, str):
        return ""
        
    # Remove diacritics and elongation
    text = strip_tashkeel(text)
    text = strip_tatweel(text)
    
    # Normalize Hamza and alef variations
    text = normalize_hamza(text)
    text = re.sub(r'[^\u0600-\u06FF\s]', '', text)  # Keep only Arabic characters
    
    # Tokenize - use a more reliable method for Arabic
    try:
        words = nltk.word_tokenize(text, language='arabic')
    except LookupError:
        # Fallback to simple whitespace tokenization if NLTK tokenization fails
        words = text.split()
    
    # Remove stopwords and short words
    filtered = [stemmer.stem(w) for w in words if w not in arabic_stopwords and len(w) > 2]
    
    return ' '.join(filtered)
```

```python
# processing my dataframe
try:
    df['Cleaned_Text'] = df['Text'].apply(preprocess_arabic)
    print(df[['Text', 'Cleaned_Text']].head())
except Exception as e:
    print(f"Error processing dataframe: {e}")
```

```python
import torch
from torch.utils.data import DataLoader, Dataset
from torch.nn.utils.rnn import pad_sequence
from torch import nn
import numpy as np
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
# Parameters for preprocessing
MAX_NB_WORDS = 20000
MAX_SEQUENCE_LENGTH = 200

# Tokenize the text and pad sequences
tokenizer = Tokenizer(num_words=MAX_NB_WORDS)
tokenizer.fit_on_texts(df['Cleaned_Text'])
X = tokenizer.texts_to_sequences(df['Cleaned_Text'])
X_pad = pad_sequences(X, maxlen=MAX_SEQUENCE_LENGTH)

# Convert labels to tensor
y = torch.tensor(df['Score'].values, dtype=torch.float32)  # For regression, use float; for classification, use long

# Convert text sequences into tensors
X_tensor = torch.tensor(X_pad, dtype=torch.long)
```

```python
class TextDataset(Dataset):
    def __init__(self, texts, labels):
        self.texts = texts
        self.labels = labels

    def __len__(self):
        return len(self.texts)

    def __getitem__(self, idx):
        return self.texts[idx], self.labels[idx]

# Create DataLoader for batching
dataset = TextDataset(X_tensor, y)
train_loader = DataLoader(dataset, batch_size=64, shuffle=True)
```

```python
class RNNModel(nn.Module):
    def __init__(self, vocab_size, embedding_dim, hidden_dim):
        super(RNNModel, self).__init__()
        self.embedding = nn.Embedding(vocab_size, embedding_dim)
        self.rnn = nn.RNN(embedding_dim, hidden_dim, batch_first=True)
        self.fc = nn.Linear(hidden_dim, 1)  # Regression output (for classification, use `softmax` and multiple units)
        
    def forward(self, x):
        x = self.embedding(x)
        out, _ = self.rnn(x)
        out = out[:, -1, :]  # Get the output of the last time step
        out = self.fc(out)
        return out
```

```python
class GRUModel(nn.Module):
    def __init__(self, vocab_size, embedding_dim, hidden_dim):
        super(GRUModel, self).__init__()
        self.embedding = nn.Embedding(vocab_size, embedding_dim)
        self.gru = nn.GRU(embedding_dim, hidden_dim, batch_first=True)
        self.fc = nn.Linear(hidden_dim, 1)  # Regression output (for classification, use `softmax` and multiple units)

    def forward(self, x):
        x = self.embedding(x)
        out, _ = self.gru(x)
        out = out[:, -1, :]  # Get the output of the last time step
        out = self.fc(out)
        return out
```

```python
class LSTMModel(nn.Module):
    def __init__(self, vocab_size, embedding_dim, hidden_dim):
        super(LSTMModel, self).__init__()
        self.embedding = nn.Embedding(vocab_size, embedding_dim)
        self.lstm = nn.LSTM(embedding_dim, hidden_dim, batch_first=True)
        self.fc = nn.Linear(hidden_dim, 1)  # Regression output (for classification, use `softmax` and multiple units)

    def forward(self, x):
        x = self.embedding(x)
        out, _ = self.lstm(x)
        out = out[:, -1, :]  # Get the output of the last time step
        out = self.fc(out)
        return out
```

```python
def train_model(model, train_loader, num_epochs=5, lr=0.001):
    criterion = nn.MSELoss()  # For regression; use `nn.CrossEntropyLoss()` for classification
    optimizer = torch.optim.Adam(model.parameters(), lr=lr)

    for epoch in range(num_epochs):
        model.train()  # Set the model to training mode
        total_loss = 0
        for texts, labels in train_loader:
            optimizer.zero_grad()
            output = model(texts)
            loss = criterion(output.squeeze(), labels)  # Remove extra dimensions
            loss.backward()
            optimizer.step()
            total_loss += loss.item()

        print(f"Epoch {epoch+1}/{num_epochs}, Loss: {total_loss / len(train_loader)}")

# put a model and train it
rnn_model = RNNModel(vocab_size=MAX_NB_WORDS, embedding_dim=128, hidden_dim=128)
train_model(rnn_model, train_loader)

lstm_model = LSTMModel(vocab_size=MAX_NB_WORDS, embedding_dim=128, hidden_dim=128)
train_model(lstm_model, train_loader)

gru_model = GRUModel(vocab_size=MAX_NB_WORDS, embedding_dim=128, hidden_dim=128)
train_model(gru_model, train_loader)
```

```python
import torch
import torch.nn as nn
import matplotlib.pyplot as plt

# Define evaluation function
def evaluate_model(model, test_loader, model_name="Model"):
    model.eval()
    total_loss = 0
    predictions = []
    targets = []

    with torch.no_grad():
        for texts, labels in test_loader:
            output = model(texts)
            output = output.squeeze()
            loss = nn.MSELoss()(output, labels)
            total_loss += loss.item()

            predictions.extend(output.tolist())
            targets.extend(labels.tolist())

    avg_loss = total_loss / len(test_loader)
    print(f"{model_name} Evaluation Loss: {avg_loss:.4f}")
    return avg_loss, predictions, targets
```

```python
losses = {}
preds_dict = {}
targets_dict = {}

models = {
    "RNN": rnn_model,
    "LSTM": lstm_model,
    "GRU": gru_model
}

for name, model in models.items():
    loss, preds, targets = evaluate_model(model, test_loader, model_name=name)
    losses[name] = loss
    preds_dict[name] = preds
    targets_dict[name] = targets
```
